```python
class CharTokenizer:
    def __init__(self):
        self.char_to_id = {}
        self.id_to_char = {}

    def fit(self,text):
        chars = sorted(set(text))
        for i, c in enumerate(chars):
            self.char_to_id[c] = i
            self.id_to_char[i] = c

        #print(self.char_to_id)
        #print(self.id_to_char)

    def encode(self,text):
        return [self.char_to_id[c] for c in text]

    def decode(self,ids):
        return "".join(
            [self.id_to_char[i] for i in ids]
        )
            
    @property
    def vocab_size(self):
        return len(self.char_to_id)

a = CharTokenizer()
a.fit("Hello, world!")
b = a.encode("Hello, world!")
a.decode(b)
b
```

```python
class WordTokenizer:
    def __init__(self, max_vocab_size=10000):
        self.max_vocab_size = max_vocab_size
        self.word_to_id = {}
        self.id_to_word = {}

    def fit(self, text):
        #split on \n and whitespace 
        words = text.lower().split()

        from collections import Counter

        counts = Counter(words)

        self.word_to_id = {
            '<PAD>' : 0,
            '<UNK>' : 1,
            '<BOS>' : 2,
            '<EOS>' : 3
        }

        self.id_to_word = {v:k for k,v in self.word_to_id.items()}

        top_words = counts.most_common(self.max_vocab_size - 4)

        for i, (word, count) in enumerate(top_words, start=4):
            self.word_to_id[word] = i
            self.id_to_word[i] = word

        #print(self.word_to_id)
        #print(self.id_to_word)

    def encode(self, text, add_special_tokens = False):
        
        words = text.lower().split()
        
        ids = [self.word_to_id.get(w,1) for w in words]

        if add_special_tokens:
            ids = [2] + ids + [3]
            
        return ids
        
        
    def decode(self, ids, skip_special_tokens=True):

        if skip_special_tokens:
            pass
        
            
    @property
    def vocab_size(self):
        return len(self.word_to_id)
```

```python
tokenizer = WordTokenizer(max_vocab_size=10)

# Train on limited text
training_text = """
The cat sat on the mat.
The cat was on the mat.
The dog sat on the mat.
"""

tokenizer.fit(training_text)
tokenizer.vocab_size

text2 = "the elephant sat" # "elephant" not in vocab!
ids2 = tokenizer.encode(text2, True)

tokenizer.decode(ids2)
```
