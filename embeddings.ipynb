```python
class CharTokenizer:
    def __init__(self):
        self.char_to_id = {}
        self.id_to_char = {}

    def fit(self,text):
        chars = sorted(set(text))
        for i, c in enumerate(chars):
            self.char_to_id[c] = i
            self.id_to_char[i] = c

        #print(self.char_to_id)
        #print(self.id_to_char)

    def encode(self,text):
        return [self.char_to_id[c] for c in text]

    def decode(self,ids):
        return "".join(
            [self.id_to_char[i] for i in ids]
        )
            
    @property
    def vocab_size(self):
        return len(self.char_to_id)

a = CharTokenizer()
a.fit("Hello, world!")
b = a.encode("Hello, world!")
a.decode(b)
b
```
